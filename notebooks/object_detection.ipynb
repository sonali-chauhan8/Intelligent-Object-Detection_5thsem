```python
from ultralytics import YOLO

model = YOLO("yolo11n.pt")

results = model.track(
    source=r"D:\B.tech\3rdyear\5thsem\Cornerstone_Project\Intelligent_Object_Detection\input\videos\video.mp4",
    stream=True,
    persist=True,
    show=False
)

vehicle_ids = set()

for result in results:
    if result.boxes.id is not None:
        track_ids = result.boxes.id.int().cpu().tolist()
        classes = result.boxes.cls.int().cpu().tolist()

        for track_id, cls in zip(track_ids, classes):
            # COCO: 2 = car, 3 = motorcycle, 5 = bus, 7 = truck
            if cls in [2, 3, 5, 7]:
                vehicle_ids.add(track_id)

print("Total unique vehicles:", len(vehicle_ids))
```
