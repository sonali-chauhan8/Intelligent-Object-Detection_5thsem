```python
# Count frames and get basic video information

import cv2

cap = cv2.VideoCapture(str(video_path))

total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
fps = cap.get(cv2.CAP_PROP_FPS)
width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

cap.release()

duration = total_frames / fps if fps > 0 else 0

print("Video Information")
print("------------------")
print("Total frames:", total_frames)
print("FPS:", fps)
print("Resolution:", width, "x", height)
print("Duration:", round(duration, 2), "seconds")
```

```python
from ultralytics import YOLO

model = YOLO("yolo11n.pt")

print("YOLO11n model loaded successfully!")
```

```python
# Real-Time Object Detection with FPS and Object Counting

import cv2
from IPython.display import display, Image, clear_output
import time

cap = cv2.VideoCapture(0)

if not cap.isOpened():
    print("Camera could not be opened.")
else:
    print("Live detection started. Interrupt the cell to stop.")

    prev_time = time.time()
    fps_values = []

    try:
        while True:
            success, frame = cap.read()

            if not success:
                print("Could not read camera frame.")
                break

            # YOLO11n object detection
            results = model(frame, imgsz=320, verbose=False)

            # Get detection result
            result = results[0]

            # Draw bounding boxes and labels
            annotated_frame = result.plot()

            # Count detected objects
            object_counts = {}

            if result.boxes is not None:
                for cls in result.boxes.cls:
                    class_name = result.names[int(cls)]
                    object_counts[class_name] = object_counts.get(class_name, 0) + 1

            # Calculate FPS
            current_time = time.time()
            instant_fps = 1 / (current_time - prev_time)
            prev_time = current_time

            # Smooth FPS using moving average
            fps_values.append(instant_fps)

            if len(fps_values) > 20:
                fps_values.pop(0)

            fps = sum(fps_values) / len(fps_values)

            # Display FPS
            cv2.putText(
                annotated_frame,
                f"FPS: {fps:.2f}",
                (20, 40),
                cv2.FONT_HERSHEY_SIMPLEX,
                1,
                (0, 255, 0),
                2
            )

            # Display object counts
            y_position = 80

            for object_name, count in object_counts.items():
                cv2.putText(
                    annotated_frame,
                    f"{object_name}: {count}",
                    (20, y_position),
                    cv2.FONT_HERSHEY_SIMPLEX,
                    0.8,
                    (0, 255, 0),
                    2
                )
                y_position += 30

            # Convert frame to JPEG
            _, buffer = cv2.imencode(".jpg", annotated_frame)

            # Display in Jupyter
            clear_output(wait=True)
            display(Image(data=buffer.tobytes()))

    except KeyboardInterrupt:
        print("Live detection stopped.")

    finally:
        cap.release()
```
